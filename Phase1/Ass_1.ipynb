```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
class Linear:
    def __init__(self, W: np.ndarray, b: np.ndarray):
        self.W = np.asarray(W, dtype=float)
        self.b = np.asarray(b, dtype=float)

    def forward(self, x: np.ndarray)-> np.ndarray:
        return x @ self.W.T + self.b
```

```python
# \hat{y} = f(x) = Wx+b
# mse_loss: L = (\hat{y}-y)^2
# dL\d\hat{y} = 2(\hat{y}-y)
# dL\dW = dL\d\hat{y} * d\hat{y}\dW = 2(\hat{y}-y)*x
# dL\db = dL\d\hat{y} * d\hat{y}\db = 2(\hat{y}-y)
```

```python
def mse_loss(y_pred: np.ndarray, y_true: np.ndarray):
    diff = y_pred - y_true
    L = float(np.mean(diff ** 2))
    dL_dy = (2.0 / diff.size) * diff
    return L, dL_dy
```

```python
W = np.array([[1.0, 2.0], [3.0, 4.0]])
b = np.array([5.0,5.0])
layer1 = Linear(W,b)

x = np.array([1.0,3.0])
y_hat = layer1.forward(x)
```

```python
y = np.array([10.0,13.0])

L, dL_dy = mse_loss(y_hat, y)
dL_dW = np.outer(dL_dy, x)
dL_db = dL_dy
print(f"dL_dW:{dL_dW}, dL_db:{dL_db}")
```

```python
import torch
import torch.nn as nn

layer_nn = nn.Linear(in_features=2, out_features=2)

with torch.no_grad():
    layer_nn.weight.copy_(torch.tensor(W))
    layer_nn.bias.copy_(torch.tensor(b))

x_torch = torch.tensor(x, requires_grad=False, dtype=torch.float)
y_true_torch = torch.tensor(y, requires_grad=False,  dtype=torch.float)

y_pred_torch = layer_nn(x_torch)
loss_fn = nn.MSELoss()
L_torch = loss_fn(y_pred_torch, y_true_torch)
L_torch.backward()

dL_dW_torch = layer_nn.weight.grad.numpy()
dL_db_torch = layer_nn.bias.grad.numpy()
```

```python
print(f"Torch: dL_dW:{dL_dW_torch}, dL_db:{dL_db_torch}")
```
